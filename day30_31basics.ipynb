```python
# File integrity checker
```

```python
import hashlib
import json

monitored_files = {}

def calculate_hash(file_path):

    try:
        with open(file_path, "r") as file:
            content = file.read()

    except (FileNotFoundError,json.JSONDecodeError):
        print("File not found")
        return

    data = content.encode()
    hash_object = hashlib.sha256(data)
    file_hash = hash_object.hexdigest()
    return file_hash

def register_file():
    file_path = input("Enter file path:  ")

    file_hash = calculate_hash(file_path)

    if file_hash is None:
        return

    monitored_files[file_path] = file_hash


    with open("hashes.json", "w") as file:
        json.dump(monitored_files, file,indent=4)

    print("The file has been registered successfully")

def verify_file():
    file_path = input("Enter file path: ")

    if file_path not in monitored_files:
        print("The file is not registered")
        return

    current_hash = calculate_hash(file_path)

    if current_hash is None:
        return

    if monitored_files[file_path] == current_hash:
        print("Integrity Verified")

    else:
        print("Warning! File has been modified")

def welcome_menu():
    print("1.Register file")
    print("2.Verify file")
    print("3.Exit")

try:
    with open("hashes.json", "r") as file:
        monitored_files = json.load(file)

except (FileNotFoundError,json.JSONDecodeError):
    monitored_files = {}

option = ""

while option != "3":
    welcome_menu()

    option = input("Enter Option:  ")

    if option == "1":
        register_file()

    elif option == "2":
        verify_file()

    elif option == "3":
        print("Goodbye")

    else:
        print("Invalid Option")
```
