```python
choice = ""
accounts = {"Farhan": 1000, "Ali" : 700, "Umar": 1500 }
account = input("Enter account name: ")

def show_menu():
    print("1.check balance")
    print("2.deposit cash")
    print("3.withdraw cash")
    print("4.exit")


def check_balance(account):
    print(f"{account}'s current balance : {accounts[account]}")

def get_amount():
    amount = int(input("Enter amount: "))
    return amount

def deposit(account):
    amount = get_amount()

    if amount > 0:
        accounts[account] += amount
        print("Transaction successful")
    else:
        print("Amount must be greater than 0")
        return 0


def withdraw(account):
    amount = get_amount()

    if amount <= 0:
        print("Amount must be greater than 0")
        return 0

    elif amount <= accounts[account]:
        accounts[account] -= amount
        print("Withdrawal successful")

    else:
        print("Insufficient balance")
        return 0

while choice != "4":
    show_menu()
    choice = input("Choose an option: ")

    if choice == "1":
        check_balance(account)

    elif choice == "2":
        deposit(account)

    elif choice == "3":
        withdraw(account)

else:
    print("Goodbye")
```

```python
## Banking system V1
```

```python
accounts = {"FARHAN": 1000, "ALI": 700, "UMAR": 1500}


def create_account():
    name = input("Enter Name: ")
    name = name.upper()

    if len(name) > 10 or len(name) < 3:
        print("For account creation minimum 3 alphabets or a maximum of 10 alphabets are allowed")
        return

    if not name.isalpha():
        print("Only alphabets are allowed")
        return

    if name in accounts:
        print("Account already exists")
        return

    accounts[name] = 0
    print("Account created")
    account_menu(name)


def login():
    attempts = 3

    while attempts > 0:
        name = input("Enter Name: ")
        name = name.upper()

        if name in accounts:
            account_menu(name)
            return

        attempts -= 1
        print(f"There is no name registered by the User named: {name}")
        print(f"Attempts remaining: {attempts}")

    print("Too many failed attempts")


def check_balance(account):
    print(f"{account}'s current balance : {accounts[account]}")


def get_amount():

    choice = ""
    while choice != "2":

        print("1. Enter amount")
        print("2. Exit")

        choice = input("Enter choice: ")

        if choice == "1":

            try:
                amount = int(input("Enter Amount: "))

                if amount > 0:
                    return amount

            except ValueError:
                print("Please enter a valid whole number")

        elif choice == "2":
            return

        else:
            print("Invalid choice")


def deposit(account):
    amount = get_amount()


    if amount is None:
        return

    elif amount > 0:
        accounts[account] += amount
        print("Transaction successful")

    else:
        print("Amount must be greater than 0")


def withdraw(account):
    amount = get_amount()

    if amount is None:
       return

    if amount <= 0:
        print("Amount must be greater than 0")

    elif amount <= accounts[account]:
        accounts[account] -= amount
        print("Withdrawal successful")

    else:
        print("Insufficient balance")


def account_menu(account):
    choice = ""

    while choice != "4":

        print("1. Check balance")
        print("2. Deposit cash")
        print("3. Withdraw cash")
        print("4. Logout")

        choice = input("Choose an option: ")

        if choice == "1":
            check_balance(account)

        elif choice == "2":
            deposit(account)

        elif choice == "3":
            withdraw(account)

        elif choice == "4":
            print("Logout")

        else:
            print("Invalid choice")


def welcome():
    print("1. Create new account")
    print("2. Login to existing account")
    print("3. Exit")


choice = ""

while choice != "3":

    welcome()
    choice = input("Choose an option: ")

    if choice == "1":
        create_account()

    elif choice == "2":
        login()

    elif choice == "3":
        print("Goodbye")

    else:
        print("Invalid choice")
```
