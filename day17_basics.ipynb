```python
student_name = "Ali"
marks = [85, 92, 78]
count_subjects = len(marks)
total = sum(marks)
average = (total/count_subjects)

print("Report Card")
print("name =", student_name)
print("subjects =",count_subjects)
print("total_marks =",total)
print("average_marks =", average)


if average >= 40:
    print("result = Passed")

else:
    print("result = Failed")
```
