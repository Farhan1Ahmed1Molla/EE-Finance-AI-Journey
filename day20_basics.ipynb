```python
Animals = {"Cat" : 5, "Dog" : 10, "Bird" : 7}

Customer = input(" Required animal =  ")


if Customer == "Dog":
    print(f"Available dogs = {Animals["Dog"]}")

elif Customer == "Cat":
    print(f"Available cats = {Animals["Cat"]}")


elif Customer == "Bird":
    print(f"Available birds = {Animals["Bird"]}")

else:
    print("Out of stock")
```

```python
Animals = {"Cat" : 5, "Dog" : 10, "Bird" : 7}

Customer = input(" Required animal =  ")


if Customer in Animals:
    print(f"There are {Animals[Customer]} available in stock")


else:
    print("Out of stock")
```

```python
Animals = {"Cat" : 5, "Dog" : 10, "Bird" : 7}

Customer = input(" Required animal =  ")


if Customer in Animals:
    print(f"{Customer} : {Animals[Customer]} Currently available")

else:
    print("Out of stock")
```

```python
attendance = {"Ali":"present","Sara":"present","John":"absent"}

call = input("Enter name:  ")


if call in attendance:
    print(f"{call} is {attendance[call]}")
```

```python
inventory = {"belt":{"stock" : 25,"price":1000},"motor":{"stock":6,"price":10000},"cable":{"stock":15,"price":2500}}
total = 0
for items in inventory:
    print(f"{items} Available : {inventory[items]["stock"]} and each costs-{inventory[items]["price"]}Rs ")
    total += inventory[items]["stock"]*inventory[items]["price"]
print(f"Total inventory cost = {total}")
```
