```python
## Elevator parts inventory management system
```

```python
import json

class Part:
    def __init__(self, name, category, price):
        self.name = name
        self.category = category
        self.price = price

class ElectricalPart(Part):
    def inspect(self):
        print(f"{self.name} has passed the Electrical inspection. ")

class MechanicalPart(Part):
    def inspect(self):
        print(f"{self.name} has passed the Mechanical inspection. ")

class SafetyPart(Part):
    def inspect(self):
        print(f"{self.name} has passed the Safety inspection. ")

class ElectronicPart(Part):
    def inspect(self):
        print(f"{self.name} has passed the Electronic inspection. ")


catalog = {"EL001": MechanicalPart("Door Roller", "Mechanical", 600),"EL002": MechanicalPart("Door Lock", "Mechanical", 2500),"EL003": ElectricalPart("Door Sensor", "Electrical", 3500),"EL004": MechanicalPart("Guide Shoe", "Mechanical", 1200),"EL005": ElectricalPart("Contactor", "Electrical", 1800),"EL006": ElectricalPart("Relay", "Electrical", 400),"EL007": ElectricalPart("Fuse", "Electrical", 150),"EL008": ElectricalPart("Circuit breaker", "Electrical", 1500),"EL009": ElectricalPart("Push Button", "Electrical", 450),"EL010": ElectricalPart("Hall Call Button", "Electrical", 1000),"EL011": ElectronicPart("Display Module", "Electronic", 2500),"EL012": SafetyPart("Emergency Battery", "Safety", 5000),"EL013": ElectronicPart("Encoder", "Electronic", 8000),"EL014": ElectronicPart("VFD", "Electronic", 80000),"EL015": MechanicalPart("Elevator Motor", "Mechanical", 150000)}
inventory = {}

def add_stock():
    for catalog_id in catalog:
        part = catalog[catalog_id]
        print(catalog_id,part.name,part.category,part.price)

    part_id = input("Enter Parts ID:  ").upper()

    if part_id not in catalog:
        print("Invalid ID")
        return

    part = catalog[part_id]
    part.inspect()

    try:
        quantity = int(input("Enter Quantity:  "))
    except ValueError:
        print("Enter a valid whole number")
        return

    if quantity <= 0:
        print(" Quantity must be greater than 0 ")
        return

    if part_id in inventory:
        inventory[part_id]["Quantity"] += quantity
        print("Stock updated successfully")

    else:
        inventory[part_id] = {"Quantity": quantity}
        print("Part added to inventory")

def view_inventory():
    if not inventory:
        print("inventory is empty ")
        return

    print("Part ID | Name | Category | Price | Quantity")
    print("-" * 50)

    for part_id in inventory:
        part = catalog[part_id]
        print(part_id,part.name,part.category,part.price,inventory[part_id]["Quantity"])

def delete_part():

    part_id = input("Enter Part ID:  ").upper()

    if part_id not in inventory:
        print("Part not found in inventory")
        return

    confirm = input("Delete this part? (Y/N):  ").upper()

    if confirm not in ("Y","N"):
        print("invalid choice")
        return

    if confirm == "N":
        print(part_id, "deletion cancelled")
        return

    if confirm == "Y":
        del inventory[part_id]
        print(part_id," successfully deleted")


def welcome_menu():
   print("1.Add Stock")
   print("2.View Inventory")
   print("3.Delete Part")
   print("4.Exit")

try:
    with open("inventory.json", "r") as file:
        inventory = json.load(file)

except (FileNotFoundError,json.JSONDecodeError):
    inventory = {}

option = ""

while option != "4":
    welcome_menu()

    option = input("Enter option: ")

    if option == "1":
        add_stock()

    elif option == "2":
        view_inventory()

    elif option == "3":
        delete_part()

    elif option == "4":
        with open("inventory.json", "w") as file:
            json.dump(inventory, file, indent=4)

        print("Goodbye")

    else:
        print("Invalid option")
```
