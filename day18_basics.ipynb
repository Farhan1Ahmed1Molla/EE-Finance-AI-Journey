```python
num1 = float(input("choose a number:  "))
operator = input("choose Operator:  ")
num2 = float(input("choose a number:  "))

def add(num1,num2):
    return num1+num2
def sub(num1,num2):
    return num1-num2
def mul(num1,num2):
    return num1*num2
def div(num1,num2):
    return num1/num2

if operator == "+":
    print(add(num1,num2))
elif operator == "-":
    print(sub(num1,num2))
elif operator == "*":
    print(mul(num1,num2))
elif operator == "/":
 try:
   print(div(num1,num2))
 except:
   print("Invalid division")
else:
    print("Invalid operator")
```
