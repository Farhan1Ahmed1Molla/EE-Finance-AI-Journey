```python
## Contact book (prototype)
```

```python
name = input("Save name = ")
phone = input("Phone number = ")

contacts = {}

contacts['Name'] = name
contacts['Phone'] = phone

search = input("Search name = ")

if search == contacts['Name']:
    print(contacts['Phone'])

else:
    print("CONTACTS NOT FOUND")
```

```python
## Contact book Version 1
```

```python
contacts = {}

choice = ""

while choice != "4":
    print("1.Add contact")
    print("2.Search contact")
    print("3.View contact")
    print("4.Exit")
    choice = input("Choose an option: ")

    if choice == "1":
        name = input("Save name = ")
        phone = input("Phone number = ")

        contacts[name] = phone

    elif choice == "2":
        search = input("Search name = ")

        if search in contacts:
            print(contacts[search])
        else:
            print("Contact not found")

    elif choice == "3":
        print(contacts)

    elif choice == "4":
        pass

    else:
        print("Invalid menu option")
```
