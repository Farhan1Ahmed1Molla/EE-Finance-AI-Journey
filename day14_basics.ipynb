```python
names = ["Ali", "Hamza", "Umar", "Yasin", "Uthman"]
attendance = [True, False, True, True, False]

for number in range(5):
    if attendance[number]:
        print(f"{number + 1}. {names[number]} - Present")
    else:
        print(f"{number + 1}. {names[number]} - Absent")
```
